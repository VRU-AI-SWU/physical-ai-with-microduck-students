# Lab 1 — ขับ Microduck ใน browser

**EP01 · Physical AI คืออะไร** · คลิปก่อนหน้า: ทฤษฎี EP01

ในคลิปทฤษฎี เราเห็นว่าหุ่นยนต์ทำงานเป็นวง **sense → think → act** และ policy ที่ทำให้ Microduck เดินได้
มาจากการเรียนรู้ใน simulator วันนี้เราจะลองขับมันเอง แล้วจดว่ามันทำอะไรได้บ้าง

simulator นี้เป็นของ Pollen Robotics ผู้สร้าง Microduck (โมเดล 3 มิติ © Pollen Robotics, CC BY-NC-SA 4.0)

## เตรียมตัว

- ใช้ **คอมพิวเตอร์ที่มีคีย์บอร์ด** และ browser รุ่นใหม่ (Chrome, Edge, Firefox หรือ Safari) ไม่ต้องติดตั้งอะไร
- เปิด https://huggingface.co/spaces/pollen-robotics/microduck-simulator แล้วกด **WADDLE IN** หรือ Enter
- เปิด notebook นี้บน Colab ไว้ข้างๆ เพื่อจดสิ่งที่เห็นในตารางท้าย notebook (ดับเบิลคลิก cell เพื่อแก้)
- Lab นี้ยังไม่ต้องเขียน code การติดตั้ง Python, Git และ Colab อยู่ใน **Lab 2**
