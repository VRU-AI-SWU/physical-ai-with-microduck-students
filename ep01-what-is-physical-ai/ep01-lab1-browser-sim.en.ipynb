# Lab 1 — Drive Microduck in the browser

**EP01 · What is Physical AI?** · Previous video: EP01 theory

In the theory video we saw that a robot works as a **sense → think → act** loop, and that the policy that makes
Microduck walk was learned in a simulator. Today we drive it ourselves and write down what it can do.

The simulator is by Pollen Robotics, the makers of Microduck (3D model © Pollen Robotics, CC BY-NC-SA 4.0).

## Setup

- Use a **computer with a keyboard** and a recent browser (Chrome, Edge, Firefox or Safari). Nothing to install.
- Open https://huggingface.co/spaces/pollen-robotics/microduck-simulator and press **WADDLE IN** or Enter.
- Open this notebook in Colab next to it, to log what you see in the table at the end (double-click a cell to edit it).
- No code in this lab yet. Setting up Python, Git and Colab is **Lab 2**.

## Experiment 3: try to make it fall

- W and D together: walk while turning
- Kick (Q) while walking
- It fell? Press **Space** to reset any time.

## Summary + assignment

**What we saw:** a key only says *what* we want; *how* (which foot to step, how far to crouch) the policy decides itself
from its sensors, fifty times a second, and walking, sitting and kicking each use their own policy.
