## Python ชุดแรก: ตัวเลขของ Microduck

ใช้ตัวเลขจริงจากคลิปทฤษฎี: สูงราว 25 cm, หนักราว 800 g, loop ควบคุม 50 ครั้งต่อวินาที

```python
height_cm = 25
mass_g = 800
control_hz = 50
dt = 1 / control_hz
print(f"one decision every {dt} s")
print(f"{control_hz * 60} decisions per minute")
```

```python
def decisions(seconds, hz=50):
    return seconds * hz
print(decisions(10))
print(decisions(0.86))
```
