## First Python: Microduck's numbers

Using the real numbers from the theory video: about 25 cm tall, about 800 g, a control loop that runs 50 times a second.

```python
height_cm = 25
mass_g = 800
control_hz = 50
dt = 1 / control_hz
print(f"one decision every {dt} s")
print(f"{control_hz * 60} decisions per minute")
```

```python
def decisions(seconds, hz=50):
    return seconds * hz
print(decisions(10))
print(decisions(0.86))
```

## Lists and loops: the joints of Microduck's left leg

These joint names come from Microduck's model (`microduck_rl`), which we'll open ourselves in Lab 3.

```python
left_leg = ["left_hip_yaw", "left_hip_roll", "left_hip_pitch",
            "left_knee", "left_ankle"]
print(len(left_leg), "joints in the left leg")
for name in left_leg:
    if "hip" in name:
        print(name)
```

## NumPy: leg angles in the standing pose

The model stores angles in **radians**, but degrees are easier for people to read · the values below are the left leg's angles in the model's STAND pose.

```python
stand_rad = np.array([0.0, -0.0873, -0.4579, -0.0049, 0.4530])
stand_deg = np.degrees(stand_rad)
print(stand_deg.round(1))
```

## Git: saving work as snapshots

The textbook (FR §4.8) explains that Git stores work as **snapshots** of all the files at one moment, and files pass through three places:

**working tree** (files we edit) → `git add` → **staging area** (what's ready) → `git commit` → **repository** (permanent history)

```python
!git --version
!git init -q microduck-notes
import os
os.chdir("microduck-notes")
!git config user.name "Student"
!git config user.email "student@example.com"
!ls -a
```
